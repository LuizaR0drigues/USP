```python
with open("piRNA-500.txt") as pos_file:
    pos_lines = pos_file.readlines()
```

```python
with open("Non-piRNA-500.txt") as neg_file:
    neg_lines = neg_file.readlines()
```

```python
new_pos_lines = []
for line in pos_lines:
    if not line.startswith(">"):
        new_line = line.replace("U", "T")
    else:
        new_line = line
    new_pos_lines.append(new_line)
```

```python
new_neg_lines = []
for line in neg_lines:
    if not line.startswith(">"):
        new_line = line.replace("U", "T")
    else:
        new_line = line
    new_neg_lines.append(new_line)
```

```python
with open("pos_samples.fa", "w") as output_pos_file:
    for line in new_pos_lines:
        output_pos_file.write(line)
```

```python
with open("neg_samples.fa", "w") as output_neg_file:
    for line in new_neg_lines:
        output_neg_file.write(line)
```
